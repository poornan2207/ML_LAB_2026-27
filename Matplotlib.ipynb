```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


x = np.array([1, 8])
y = np.array([3, 10])

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.show()



x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.figure(figsize=(8, 5))
plt.plot(
    x, y,
    color='blue',
    linestyle=':',
    marker='s',
    linewidth=3,
    label='My Data Series'
)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Chart')
plt.legend()
plt.grid(True)
plt.show()



categories = ['A', 'B', 'C', 'D', 'E']
values = [5, 2, 3, 4, 6]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color='green')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart')
plt.show()



data = [
    12, 15, 13, 17, 16, 16, 19, 12,
    14, 16, 14, 17, 18, 18, 15, 14
]

plt.figure(figsize=(8, 5))
plt.hist(data, bins=16, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Simple Histogram Example')
plt.show()



x = [1, 2, 3, 4, 5]
y = [5, 7, 4, 6, 8]

plt.figure(figsize=(8, 5))
plt.scatter(x, y)
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.title('Simple Scatter Plot Example')
plt.show()



labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]

plt.figure(figsize=(8, 5))
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=140
)
plt.title('Simple Pie Chart Example')
plt.axis('equal')
plt.show()



data = [7, 8, 5, 6, 7, 9, 12, 5, 7, 8, 6]

plt.figure(figsize=(8, 5))
plt.boxplot(data)
plt.title('Simple Box Plot Example')
plt.ylabel('Values')
plt.show()



data = np.random.rand(10, 10)

plt.figure(figsize=(8, 5))
plt.imshow(data, cmap='viridis', interpolation='nearest')
plt.colorbar()
plt.title('Simple Heatmap Example')
plt.show()
```
